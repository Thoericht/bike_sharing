## Import data

```python
# imports

import zipfile
from pathlib import Path

import pandas as pd
import requests

pd.set_option(
    'display.float_format',
    '{:.2f}'.format
)

# save data 

project_dir = Path.cwd().resolve()

if not (project_dir / "01_data").exists() and (project_dir.parent / "01_data").exists():
    project_dir = project_dir.parent

output_dir = project_dir / "01_data" / "01_raw_data"
output_dir.mkdir(parents=True, exist_ok=True)


# download data

URL = 'https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip'
zip_path = output_dir / 'bike_sharing_dataset.zip'

response = requests.get(URL)
response.raise_for_status()

with open(zip_path, "wb") as f:
    f.write(response.content)

print(f'ZIP file saved to: {zip_path}')


# extract hour.csv

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extract('hour.csv', output_dir)

print(f"hour.csv saved to: {output_dir / 'hour.csv'}")
```
